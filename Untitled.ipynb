```python
from bs4 import BeautifulSoup
import pandas as pd

import requests
import re
from datetime import date, datetime
import os
from pathlib import Path
import sys



from utils.logger import logger



APP_DIR = Path.cwd()

DATE = datetime.now().strftime("%Y-%m-%d")

bronze_DIR = APP_DIR / "Bronze" / DATE
silver_DIR = APP_DIR / "Silver" / DATE


bronze_DIR.mkdir(parents=True, exist_ok=True)
silver_DIR.mkdir(parents=True, exist_ok=True)
```

###BRONZE LAYER

```python
logger.info("Bronze pipeline started successfully")

url = 'https://companiesmarketcap.com/'
header = { 'User-Agent' : 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.0.0 Safari/537.36'}

logger.info(f"Sending requests to {url}")
response = requests.get(url , headers =header)

if response.status_code == 200:
    logger.info("Response status code success - 200")
else:
    logger.warn(f"Response status code failed - {response.status_code}")
    sys.exit(1)
```

```python
soup = BeautifulSoup(response.text)
```

```python
ranking_bar = soup.find('div', class_='ranking-bar')
filter_tags = ranking_bar.find_all('a')

#get pages links
filter_links = [link.get('href') for link in filter_tags if link.get('href') != '#']

#get relative urls
urls = list(map(lambda x :url[:-1] + x,filter_links))

if len(urls) != 0:
    logger.info(f"Inner pages url extracted , Length :{len(urls)}")
```

```python


logger.info("Data srcaping Started")
for idx,url in enumerate(urls):

    logger.info(f"{idx+1} - url")

    try:
        response = requests.get(url, headers = header, timeout= 30)
    
        if response.status_code !=200:
            logger.warn("Request failed")
            continue
            
    except requests.exceptions.Timeout:
        logger.warn('Request timeout error')

    soup = BeautifulSoup(response.text)
    soup.find_all('h1')[0].get_text()

    stats_bar = soup.find('div' , class_='category-stats-bar')
    stats= stats_bar.get_text().strip().split('\n')

    for stat in stats:
        heading = stat.split(':')[0]
        value = stat.split(':')[1]

        print(heading ,':' , value)

    file_name = 'companies_' + stats[-1].split(':')[0].replace(' ','_').replace('/','_')

    a_tags = stats_bar.find_all('a' , class_='tooltip-title')
    download_file = [link.get('href') for link in a_tags if 'download' in link.get('href')]

    if len(download_file) == 0:
        logger.debug("Download file option not available")
        continue

    download_url = url + download_file[0]

    logger.info("Download data Request send")

    response = requests.get(download_url, timeout=30 , headers= header)
    
    response.raise_for_status()
    file_path = os.path.join(bronze_DIR, f"{file_name}.csv")

    logger.info(f"Status Code{response.status_code} - {file_path}")

    #idempotent
    if os.path.exists(file_path):
        logger.info(f"File already exists, skipping: {file_path}")
        continue 

    logger.info(f"Downloading file - {file_path}")

    with open(file_path,'wb') as f:
        f.write(response.content)

logger.info("Bronze pipeline completed successfully")
```

###SILVER LAYER

```python
# Silver:

# Clean, typed, standardized, validated data.

# For your CompaniesMarketCap data, we'll do:

# Read Bronze CSV
# Clean column names
# Clean strings
# Convert data types
# Handle nulls
# Remove duplicates
# Add snapshot_date
# Validate
# Save as Parquet
```

```python
def transform_data(df , filepath):
    df.head()
    df.info()
    
    df.isnull().sum()
    
    df [df.isna().any(axis= 1)]
    
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace(r"[()]", "", regex=True)
    )
    
    df = df.rename(columns={'name':'company_name'})
    
    df.duplicated().sum()
    df.drop_duplicates(inplace=True)
    
    #clean string columns_tags
    string_columns = ['company_name', 'symbol', 'country']
    
    for col in string_columns:
        df[col] = df[col].astype('string').str.strip()
    
    
    #convert abnormal values to nan
    numeric_columns = [
        col for col in df.columns
        if col not in string_columns
    ]
    
    for col in numeric_columns:
        df[col] = pd.to_numeric(
            df[col],
            errors='coerce'
        )
    
    logger.info("Snapshots added")
    df['snapshot_time'] = pd.to_datetime(DATE)
    
    if df['company_name'].isna().sum():
        logger.warning("Some companies have missing names")
    
    if df['rank'].isna().sum():
        logger.warning("Some companies have missing ranks")
    
    if df['symbol'].isna().sum():
        logger.warning("Some companies have missing symbol")
    
    if (df["rank"] <= 0).any():
        logger.warning("Invalid rank detected")

    df.to_parquet(filepath, engine='pyarrow')
    logger.info("Saved as parquet file")
        
    
        
```

```python
logger.info("Silver pipeline started successfully")

for file in os.listdir(bronze_DIR):
    if file.endswith('.csv'):
        logger.info(f"Processing {file}")
        df = pd.read_csv((bronze_DIR / file))
        transform_data(df , (silver_DIR / file))
        
        
logger.info("Silver transformation pipeline ended sucessfully")
```
